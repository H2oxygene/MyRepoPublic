# Génération de texte avec Transformers

Exemple minimal exécutable dans Google Colab.

```python
!pip -q install transformers

from transformers import pipeline

generator = pipeline(
    'text-generation',
    model='distilgpt2',
    tokenizer='distilgpt2'
)

prompt = 'Écris une courte introduction sur l’intelligence artificielle :'
result = generator(
    prompt,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.9,
    num_return_sequences=1
)
print(result[0]['generated_text'])
```
